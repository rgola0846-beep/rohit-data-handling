# Student Academic Performance Analytics Dashboard
Complete project using Python, Pandas, NumPy and Plotly.
Dataset: 30 synthetic student records.

## Objective
Analyze study hours, attendance, marks, program and assessment date.

```python
# Import libraries and create the dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

np.random.seed(42)
programs = ['BCA', 'BBA', 'B.Tech', 'B.Sc']
df = pd.DataFrame({
 'Study Hours': np.random.randint(1,9,30),
 'Attendance': np.random.randint(55,101,30),
 'Program': [programs[i%4] for i in range(30)],
 'Assessment Date': pd.date_range('2026-01-01', periods=30, freq='7D')
})
df['Marks'] = (25 + df['Study Hours']*5 + df['Attendance']*0.25 + np.random.normal(0,7,30)).round().clip(0,100).astype(int)
display(df)
df.to_csv('Student_Academic_Data.csv', index=False)

# Inspection
print(df.shape)
df.info()
display(df.describe())
print(df['Program'].value_counts())

# Correlation
display(df[['Study Hours','Attendance','Marks']].corr())

# Groupby
program_analysis = df.groupby('Program').agg(
 Average_Marks=('Marks','mean'),
 Average_Attendance=('Attendance','mean'),
 Average_Study_Hours=('Study Hours','mean'),
 Student_Count=('Marks','count')
).round(2)
display(program_analysis)

# Visualizations
fig = px.scatter(df, x='Study Hours', y='Marks', color='Program', title='Higher Study Hours Are Associated With Marks')
fig.show()
fig = px.bar(program_analysis, x=program_analysis.index, y='Average_Marks', title='Average Marks Across Academic Programs')
fig.show()
fig = px.scatter(df, x='Attendance', y='Marks', color='Program', title='Attendance and Academic Performance')
fig.show()

# Interactive dashboard
dashboard = make_subplots(rows=2, cols=2, subplot_titles=(
 'Study Hours vs Marks','Attendance vs Marks',
 'Average Marks by Program','Average Study Hours by Program'))
for p in programs:
 s = df[df['Program']==p]
 dashboard.add_trace(go.Scatter(x=s['Study Hours'], y=s['Marks'], mode='markers', name=p), row=1, col=1)
dashboard.add_trace(go.Scatter(x=df['Attendance'], y=df['Marks'], mode='markers', name='Attendance'), row=1, col=2)
dashboard.add_trace(go.Bar(x=program_analysis.index, y=program_analysis['Average_Marks'], name='Average Marks'), row=2, col=1)
dashboard.add_trace(go.Bar(x=program_analysis.index, y=program_analysis['Average_Study_Hours'], name='Study Hours'), row=2, col=2)
dashboard.update_layout(height=850, title_text='Student Academic Performance Analytics Dashboard')
dashboard.show()
dashboard.write_html('Student_Academic_Dashboard.html')
```
